```python
# Week 1 Project: Simple Calculator

#This notebook takes two numbers and performs addition, subtraction, multiplication, and #division.
```

```python
print("===== MY SIMPLE CALCULATOR =====")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\nChoose an operation:")
print("+  Addition")
print("-  Subtraction")
print("*  Multiplication")
print("/  Division")

choice = input("Enter your choice (+, -, *, /): ")

if choice == "+":
    answer = num1 + num2
    print("Answer =", answer)

elif choice == "-":
    answer = num1 - num2
    print("Answer =", answer)

elif choice == "*":
    answer = num1 * num2
    print("Answer =", answer)

elif choice == "/":
    if num2 == 0:
        print("Error: Cannot divide by zero.")
    else:
        answer = num1 / num2
        print("Answer =", answer)

else:
    print("Invalid operation!")
```
